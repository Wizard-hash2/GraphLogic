# Computing the gradients of the loss with respect to trainable variables

lets try to perform a simple , where we will apply above:
z = wx + y
summation of loss = (y-z)**2

```python
w = torch.tensor(1.0, requires_grad= True)
b = torch.tensor(0.5, requires_grad= True)
x = torch.tensor([1.4])
y = torch.tensor([2.1])
z = torch.add(torch.mul(w, x), b)
print('dL/dw : ', w.grad)
print('dL/db : ', b.grad)
loss = (y-z).pow(2).sum()
loss.backward()
print('dL/dw : ', w.grad)
print('dL/db : ', b.grad)

#note we compute z using forward pass in n n
# then we use backward method on the loss to compute derivation of loss with respect to w and b
```

in above note we compute z using forward pass in nn then we use backward method on the loss to compute derivation of loss with respect to w and b
